```python
Build a Python stock analysis app called "ai-investor-panel" with a multi-agent 
architecture where each agent analyzes a stock through the lens of a famous investor.

## Architecture
- CLI entry point: `python main.py --ticker AAPL` (support multiple tickers 
  comma-separated, e.g. --ticker AAPL,NVDA,TSLA)
- Use yfinance for market data (price history, fundamentals, financials, 
  valuation ratios). No paid APIs.
- Each agent is a separate module in an /agents folder with a shared base class 
  that takes the fetched data and returns: signal (bullish/neutral/bearish), 
  confidence (0-100), and reasoning (2-3 sentences in that investor's voice).

## Agents to implement
1. warren_buffett: Focuses on durable competitive moats, owner earnings, ROE 
   consistency, low debt, margin of safety vs intrinsic value (DCF with 
   conservative growth assumptions). Skeptical of anything he can't understand.
2. charlie_munger: Quality of business over price, mental models, avoids 
   stupidity rather than seeking brilliance. Penalizes complexity and hype.
3. cathie_wood: Disruptive innovation focus — revenue growth rate, TAM 
   expansion, R&D intensity, exponential adoption curves. Tolerates losses 
   if growth trajectory is steep. Uses 5-year forward price targets.
4. ben_graham: Deep value — net-net working capital, P/E and P/B screens, 
   Graham Number, dividend record, balance sheet strength.
5. michael_burry: Contrarian deep value — free cash flow yield, insider 
   activity, looks for what the market is missing or overhyping. 
   Naturally skeptical of crowded trades.
6. peter_lynch: Growth at a reasonable price — PEG ratio, earnings momentum, 
   category of stock (stalwart, fast grower, cyclical, turnaround).
7. technical_analyst: Trend (50/200 SMA), RSI, MACD, volume patterns, 
   support/resistance. Purely price-based.
8. risk_manager: Position sizing, volatility (beta, max drawdown), 
   correlation warnings, flags concentration risk.

## Synthesis
- portfolio_manager agent: takes all agent outputs, weighs them, and produces 
  a final verdict per ticker: overall signal, conviction score, key bull case, 
  key bear case, and what would change the thesis.
- Agents can disagree — surface the disagreement explicitly instead of 
  averaging it away.

## Output
- Rich terminal output using the `rich` library: one table per ticker with 
  each agent's signal/confidence/one-line reasoning, then the portfolio 
  manager's summary panel.
- Also write a markdown report to /reports/{ticker}_{date}.md.

## Code quality
- Type hints, dataclasses for signals, graceful handling of missing data 
  (e.g. no dividends, negative earnings), retry logic on data fetches.
- Include a README with setup instructions and a prominent disclaimer that 
  this is for educational purposes and not financial advice.
- Add a --show-work flag that prints the raw metrics each agent used.

Start by scaffolding the project structure, then implement the base agent 
and data fetcher, then agents one at a time, testing with AAPL as you go.
```
